```python
def add_missingness_indicators(df, cols_with_nan):
    df = df.copy()

    for col in cols_with_nan:
        df[f"{col}_missing"] = df[col].isna().astype("float32")
        df[col] = df[col].fillna(0)

    return df
```

```python
# 1. Identify missing columns using training data only
missing_cols = [col for col in FEATURE_COLS if train_df[col].isna().any()]

# 2. Add missingness indicators and impute
train_df = add_missingness_indicators(train_df, missing_cols)
val_df = add_missingness_indicators(val_df, missing_cols)
test_df = add_missingness_indicators(test_df, missing_cols)
```

```python
# 3. Define model inputs and targets separately
X_train = train_df[FEATURE_COLS + [f"{col}_missing" for col in missing_cols]]
X_val = val_df[X_train.columns]
X_test = test_df[X_train.columns]

TARGET_COL = "is_churn"

y_train = train_df[TARGET_COL]
y_val = val_df[TARGET_COL]
y_test = test_df[TARGET_COL]
```

```python
# onehot registered_ia
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
```

```python
CONTINUOUS_COLS = [
    "n_txn",
    "n_cancel",
    "last_actual_amount_paid",
    "last_payment_plan_days",
    "last_plan_list_price",
    "last_discount",
    "cancel_rate",
    "days_since_last_txn",
    "log_days",
    "avg_daily_secs",
    "total_secs_sum",
    "avg_completion_ratio",
    "avg_daily_unq_songs",
    "days_since_last_log",
    "tenure_days",
]
```

```python
# 4. Define preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ('encoder', OneHotEncoder(handle_unknown='ignore'), ['registered_via']),
        ('scale', StandardScaler(), CONTINUOUS_COLS)
    ],
    remainder='passthrough'
)

# 5. Fit on training data only
X_train_processed = preprocessor.fit_transform(X_train)

# Transform validation and test using fitted parameters
X_val_processed = preprocessor.transform(X_val)
X_test_processed = preprocessor.transform(X_test)

# 6. Inspect resulting shapes
print("Train:", X_train_processed.shape)
print("Validation:", X_val_processed.shape)
print("Test:", X_test_processed.shape)

print(type(X_train_processed))
print(X_train_processed.shape)
print(X_train_processed.dtype)
print("NaNs:", pd.isna(X_train_processed).sum())
```
